```python
feature_columns = [
    "Applied_Voltage_kV",
    "Load_Current_A",
    "Ambient_Temperature_C",
    "Test_Duration_min",
    "Sensor_S1",
    "Sensor_S2",
    "Sensor_S3",
    "Sensor_S4"
]

train[feature_columns] = train[feature_columns].apply(
    pd.to_numeric, errors="coerce"
)

test[feature_columns] = test[feature_columns].apply(
    pd.to_numeric, errors="coerce"
)
```

```python
for col in feature_columns:
    median_value = train[col].median()

    train[col] = train[col].fillna(median_value)
    test[col] = test[col].fillna(median_value)
```

```python
# ============================================================
# PREPROCESSING - STEP 1
# Create a separate copy of the original training data
# ============================================================

cleaned_train = train.copy()

print("Original training shape:", train.shape)
print("Working preprocessing shape:", cleaned_train.shape)
```

```python
# ==========================================
# STEP 2 - Reload ORIGINAL training data
# ==========================================

train_raw = pd.read_excel(
    file_path,
    sheet_name="Training_Data"
)

# Make a fresh working copy
cleaned_train = train_raw.copy()

sensor_columns = [
    "Sensor_S1",
    "Sensor_S2",
    "Sensor_S3",
    "Sensor_S4"
]

# Check original missing values
print("Original missing values:")
print(cleaned_train[sensor_columns].isna().sum())

# Create missing-value indicators BEFORE filling anything
for col in sensor_columns:
    cleaned_train[col + "_missing"] = (
        cleaned_train[col].isna().astype(int)
    )

print("\nMissing indicators:")
print(
    cleaned_train[
        [col + "_missing" for col in sensor_columns]
    ].sum()
)
```

```python
# ============================================================
# PREPROCESSING - STEP 1
# Reload the ORIGINAL raw training data
# ============================================================

train_raw = pd.read_excel(
    file_path,
    sheet_name="Training_Data"
)

print("Raw training data shape:", train_raw.shape)
print("\nMissing values in raw sensor data:")

print(
    train_raw[
        ["Sensor_S1", "Sensor_S2", "Sensor_S3", "Sensor_S4"]
    ].isna().sum()
)
```

```python
# ============================================================
# PREPROCESSING - STEP 2
# Create preprocessing dataframe
# ============================================================

cleaned_train = train_raw.copy()

print("Rows:", len(cleaned_train))
print("Columns:", len(cleaned_train.columns))
```

```python
# ============================================================
# PREPROCESSING - STEP 3
# Define columns
# ============================================================

feature_columns = [
    "Applied_Voltage_kV",
    "Load_Current_A",
    "Ambient_Temperature_C",
    "Test_Duration_min",
    "Sensor_S1",
    "Sensor_S2",
    "Sensor_S3",
    "Sensor_S4"
]

sensor_columns = [
    "Sensor_S1",
    "Sensor_S2",
    "Sensor_S3",
    "Sensor_S4"
]

id_column = "Test_ID"

regression_target = "Reference_Parameter"

classification_target = "Validity_Label"

print("Input features:")
print(feature_columns)

print("\nRegression target:", regression_target)
print("Classification target:", classification_target)
print("ID column:", id_column)
```

```python
# ============================================================
# PREPROCESSING - STEP 4
# Create a SEPARATE copy for missing-value indicators
# ============================================================

indicator_train = train_raw.copy()

for col in sensor_columns:
    indicator_train[col + "_missing"] = (
        indicator_train[col].isna().astype(int)
    )

print("Missing-value indicators:")
print(
    indicator_train[
        [col + "_missing" for col in sensor_columns]
    ].sum()
)
```

```python
# ============================================================
# PREPROCESSING - STEP 5
# Save dataset with missing-value indicators
# ============================================================

indicator_train.to_csv(
    "cleaned_training_with_missing_indicators.csv",
    index=False
)

print("File saved successfully.")
print("Shape:", indicator_train.shape)
```

```python
# ============================================================
# PREPROCESSING - STEP 6
# Calculate median values for imputation
# ============================================================

medians = {}

for col in feature_columns:
    medians[col] = train_raw[col].median()

print("Median values:")

for col, value in medians.items():
    print(f"{col}: {value}")
```

```python
## ============================================================
# PREPROCESSING - STEP 7
# Median imputation
# ============================================================

for col in feature_columns:
    cleaned_train[col] = cleaned_train[col].fillna(medians[col])

print("Missing values after imputation:")

print(
    cleaned_train[feature_columns].isna().sum()
)
```

```python
# ============================================================
# PREPROCESSING - STEP 8
# Verify row count
# ============================================================

print("Original rows:", len(train_raw))
print("Cleaned rows:", len(cleaned_train))

assert len(train_raw) == len(cleaned_train)

print("\nSUCCESS: All 1000 rows have been retained.")
```

```python
# ============================================================
# PREPROCESSING - STEP 9
# Verify ID and target columns
# ============================================================

assert cleaned_train["Test_ID"].equals(
    train_raw["Test_ID"]
)

assert cleaned_train["Reference_Parameter"].equals(
    train_raw["Reference_Parameter"]
)

assert cleaned_train["Validity_Label"].equals(
    train_raw["Validity_Label"]
)

print("Test_ID: unchanged")
print("Reference_Parameter: unchanged")
print("Validity_Label: unchanged")
```

```python
# ============================================================
# PREPROCESSING - STEP 10
# Final verification
# ============================================================

print("Final shape:", cleaned_train.shape)

print("\nMissing values:")
print(cleaned_train.isna().sum())

print("\nFinal columns:")
print(cleaned_train.columns.tolist())
```

```python
# ============================================================
# PREPROCESSING - STEP 11
# Save cleaned training dataset
# ============================================================

cleaned_train.to_csv(
    "cleaned_training_data.csv",
    index=False
)

print("SUCCESS!")
print("Saved: cleaned_training_data.csv")
print("Shape:", cleaned_train.shape)
```

```python
# ============================================================
# PREPROCESSING - STEP 15
# Create preprocessing README
# ============================================================

readme_text = """
CPRI HACKATHON - DATA PREPROCESSING

RAW DATA
The original raw Excel file was not modified.

TRAINING DATA
1000 training records were retained.
No rows were removed.

INPUT FEATURES
1. Applied_Voltage_kV
2. Load_Current_A
3. Ambient_Temperature_C
4. Test_Duration_min
5. Sensor_S1
6. Sensor_S2
7. Sensor_S3
8. Sensor_S4

TARGET COLUMNS
Regression target:
Reference_Parameter

Classification target:
Validity_Label

IDENTIFIER
Test_ID

Test_ID is retained for identification and final output,
but should not be used as a model input feature.

MISSING VALUES
Missing values were present in Sensor_S1, Sensor_S2,
Sensor_S3 and Sensor_S4.

Missing numerical values were replaced using median
imputation.

Median values were calculated from the training data.

No rows were deleted because of missing values.

OUTLIERS
Unusual sensor values were not automatically removed.

SCALING
No feature scaling was performed during base preprocessing.

FEATURE ENGINEERING
No engineered features were added during preprocessing.
Feature engineering will be performed separately.

DATASET VERSIONS

1. cleaned_training_data.csv
   Basic cleaned dataset with 11 columns.

2. cleaned_training_with_missing_indicators_imputed.csv
   Cleaned dataset with 4 additional missing-value
   indicator columns.

MISSING-VALUE INDICATORS

Sensor_S1_missing = 1 if Sensor_S1 was originally missing,
otherwise 0.

Sensor_S2_missing = 1 if Sensor_S2 was originally missing,
otherwise 0.

Sensor_S3_missing = 1 if Sensor_S3 was originally missing,
otherwise 0.

Sensor_S4_missing = 1 if Sensor_S4 was originally missing,
otherwise 0.

The sensor values are median-imputed while the indicators
preserve information about whether the original value was
missing.

IMPORTANT
Test_ID should not be used as a model feature.
Reference_Parameter and Validity_Label are target columns.

The original raw Excel dataset remains unchanged.
"""

with open("preprocessing_README.txt", "w") as f:
    f.write(readme_text)

print("SUCCESS!")
print("Saved: preprocessing_README.txt")
```
